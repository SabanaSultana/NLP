# Word Stemming

```python
from nltk.stem import PorterStemmer

stemmer=PorterStemmer()

words=['eat', 'eats', 'eaten', 'finally', 'final', 'congratulation', 'congo', 'congrats', 'history', 'historical']

for word in words:
  print(word, ":", stemmer.stem(word))
```

```python
from nltk.stem import RegexpStemmer

regexp_stemmer=RegexpStemmer('ing$|e|es$|$ing|^sa')   ## $ after literal means remove suffix, $ before literal not workful, ^ for prefix, only literal represent anywhere in the text

print(regexp_stemmer.stem('eating'))
print(regexp_stemmer.stem('ingeat'))
print(regexp_stemmer.stem('ingeated'))
print(regexp_stemmer.stem('sabana'))
```

```python
from nltk.stem import SnowballStemmer
stemming=SnowballStemmer('english')
for word in words:
  print(word, ":", stemming.stem(word))


```

# Name Entity Recognition

```python
import nltk
nltk.download('punkt') # For word tokenization
nltk.download('averaged_perceptron_tagger') # For POS tagging
nltk.download('maxent_ne_chunker') # For Named Entity Recognition
nltk.download('words') # For the words corpus used by the chunker

sentence = "Barack Obama was born in Hawaii and served as the 44th President of the United States."

# Step 1: Tokenize the sentence into words
words = nltk.word_tokenize(sentence)
print(f"Tokenized words: {words}\n")

# Step 2: Perform Part-of-Speech (POS) tagging
tagged_words = nltk.pos_tag(words)
print(f"POS-tagged words: {tagged_words}\n")

# Step 3: Perform Named Entity Recognition (NER)
named_entities = nltk.ne_chunk(tagged_words)
print("Named Entities:\n")
print(named_entities)
```

As you can see, the `nltk.ne_chunk` function correctly identifies "Barack Obama" as a PERSON, "Hawaii" as a GPE (Geo-Political Entity), and "United States" as a GPE, all thanks to the preceding POS tagging.

# Bag of Words

```python
import nltk
import re
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

ps=PorterStemmer()
```

```python
sentences = [
    "NLP is AMAZING!!! 😃",
    "I bought 3 apples, 2 bananas, and 1 mango.",
    "Email me at example123@gmail.com.",
    "The price of Bitcoin in 2025 was $45000.",
    "Running, runner, and ran are forms of the same word.",
    "Visit https://example.com for more info.",
    "AI & ML are the future of technology.",
    "   Extra   spaces should be removed.   ",
    "COVID-19 changed the world in 2020.",
    "Is this sentence properly tokenized?"
]
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

ps = PorterStemmer()
stop_words = set(stopwords.words('english'))   # ✅ Store once (fast lookup)

cleaned_sentences = []

for sent in sentences:

    # 1️⃣ Lowercase
    sent = sent.lower()

    # 2️⃣ Remove non-letters
    sent = re.sub('[^a-z]', ' ', sent)

    # 3️⃣ Tokenize
    words = sent.split()

    # print(words, '...')

    # 4️⃣ Remove stopwords + Stemming
    words = [ps.stem(word) for word in words if word not in stop_words]

    print(words)

    cleaned_sentences.append(" ".join(words))

print("\nFinal Cleaned Sentences:\n", cleaned_sentences)
```

```python
### Now do the BOW vectorization
## Create the Bag OF Words model
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
cv=CountVectorizer(max_features=10,binary=True)
```

```python
X=cv.fit_transform(cleaned_sentences).toarray()
```
