# Word Stemming

```python
from nltk.stem import PorterStemmer

stemmer=PorterStemmer()

words=['eat', 'eats', 'eaten', 'finally', 'final', 'congratulation', 'congo', 'congrats', 'history', 'historical']

for word in words:
  print(word, ":", stemmer.stem(word))
```

```python
from nltk.stem import RegexpStemmer

regexp_stemmer=RegexpStemmer('ing$|e|es$|$ing|^sa')   ## $ after literal means remove suffix, $ before literal not workful, ^ for prefix, only literal represent anywhere in the text

print(regexp_stemmer.stem('eating'))
print(regexp_stemmer.stem('ingeat'))
print(regexp_stemmer.stem('ingeated'))
print(regexp_stemmer.stem('sabana'))
```

```python
from nltk.stem import SnowballStemmer
stemming=SnowballStemmer('english')
for word in words:
  print(word, ":", stemming.stem(word))


```

```python
from nltk.stem import SnowballStemmer

# Initialize SnowballStemmer for German
german_stemmer = SnowballStemmer('german')

# List of German words to stem
german_words = ['häuser', 'mädchen', 'geliebt', 'geht', 'gegangen', 'spielen', 'gespielt']

print("Stemming German words:")
for word in german_words:
  print(f"{word} : {german_stemmer.stem(word)}")
```

```python
from nltk.corpus import stopwords
stopwords.words( 'english')

para=""" Natural Language Processing (NLP) is a fascinating field at the intersection of computer science, artificial intelligence, and linguistics. It focuses on enabling computers to understand, interpret, and generate human language in a valuable way. This technology powers many everyday applications, from virtual assistants and translation tools to spam filters and sentiment analysis systems. Developing effective NLP models requires a deep understanding of linguistic structures and statistical methods. Ultimately, NLP aims to bridge the communication gap between humans and machines, making technology more intuitive and accessible for everyone """

sentences=sent_tokenize(para)

print(type(sentences))
```

```python
sentences
```

```python
for i in range(len(sentences)):
  words=word_tokenize(sentences[i])

  words=[lemmatizer.lemmatize(word) for word in words if word not in stopwords.words('english')]
  sentences[i]=' '.join(words)
```

```python
for sent in sentences:
  print(sent)
```
